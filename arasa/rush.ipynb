```python
import numpy as np
from tqdm import tqdm
import random
```

```python
def simulate(spec):
    ball = []
    for i in range(100000):
        ball.append(spec())
    print(np.array(ball).mean(), np.array(ball).std())
```

```python
def eva():
    ball = 0
    # 81%継続1500発ループ
    while True:
        if random.random() < 0.81:
            ball += 1500
        else:
            break
    return ball
```

```python
def rezero():
    ball = 0
    while True:
        x = random.random()
        if x <= 0.7651256:
            furiwake = random.random()
            if furiwake < 0.14:
                ball += 0
            elif furiwake < 0.2:
                ball += 300
            elif furiwake < 0.75:
                ball += 1500
            else:
                ball += 3000
        else:
            break
    return ball
```

```python
def ghoul():
    ball = 0
    # 75%継続3000発ループ
    while True:
        if random.random() < 0.745:
            ball += 3000
        else:
            break
    return ball
```

```python
def karakuri():
    ball = 0
    # 80%継続2000発ループ
    while True:
        if random.random() < 0.8:
            furiwake = random.random()
            if furiwake < 0.1:
                ball += 0
            elif furiwake < 0.5:
                ball += 1500
            elif furiwake < 0.83:
                ball += 3000
            elif furiwake < 0.97:
                ball += 4500
            else:
                ball += 6000


        else:
            break
    return ball
```

```python
def madoka():
    ball = 0
    # 75%継続3000発ループ
    while True:
        if random.random() < 0.865:
            ball += 1500
        else:
            break
    return ball
```
